## Part 2: Data Summarization

```python
# Total "Gross" of all top 1000 movies
imdb['Gross'].sum()
```

```python
# Average "No_of_Votes" of all movies
imdb['No_of_Votes'].mean()
```

```python
# Count movies in each decade (e.g., ..., 1980, 1990, 2000, 2010, 2020)
# Sort decades by the number of movies in descending order
imdb['Decade'].value_counts().sort_values(ascending=False)
```

```python
# Count movies by different directors.
# Show the top 10 directors with the most movies in this list

imdb['Director'].value_counts().head(10)
```

```python
# For each director, calculate the number of movies, average IMDB_Rating, and total gross.
# Rank them in descending order of total gross
# Show the top 10 directors with the highest total gross

(
    imdb.groupby('Director').agg({
        'Series_Title' : 'count',
        'IMDB_Rating' : 'mean',
        'Gross' : 'sum'
    }).sort_values('Gross', ascending=False).head(10)
)
```

```python
# Group movies by decade and director
# In each group (i.e., for each decade and each director),
# calculate the number of movies and average IMDB rating.
# Sort in descending order of movie count

(
    imdb.groupby(['Decade', 'Director'])
    .agg({'Series_Title' : 'count',
          'IMDB_Rating' : 'mean'
    }).sort_values('Series_Title', ascending=False)
)
```

```python
# Find the top 3 highest rated movies of each year since 2010.
# Show their released year, ranking, title, and IMDB_Rating.
# Sort them in descending order of year and ascending order of ranking
imdb['Ranking'] = imdb.groupby('Released_Year')['IMDB_Rating'].rank(method='first', ascending=False)


(
imdb[imdb['Released_Year'] >= 2010]
.sort_values(['Released_Year', 'IMDB_Rating'], ascending=[False, False])
.groupby('Released_Year')
.head(3)
[['Released_Year', 'Ranking', 'Series_Title', 'IMDB_Rating']]
  .sort_values(['Released_Year', 'Ranking'], ascending=[False, True])
)
```

```python
# Find all directors whose movies appeared in at least five different decades.
# Your result should include: director, decade, and the number of movies in the decade.

Count_director = imdb.groupby(['Director', 'Decade']).agg({
    'Series_Title' : 'count'
}).reset_index()

Count_director = Count_director.rename(
    columns={'Series_Title' : "Movie_Count"}
)

Director_Decade_Count = Count_director.groupby('Director')['Decade'].nunique()

Five_Decade_Director_Count = Director_Decade_Count[Director_Decade_Count >= 5].index

Count_director[Count_director['Director'].isin(Five_Decade_Director_Count)]
```
